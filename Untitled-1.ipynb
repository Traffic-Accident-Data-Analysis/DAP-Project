```python
import pandas as pd
import os

# --- 1. CẤU HÌNH ĐƯỜNG DẪN ---
file_path = os.path.join('data', 'US_Accidents_March23.csv')

# --- 2. DANH SÁCH CỘT CẦN GIỮ LẠI ---
cols_to_keep = [
    # Mục tiêu & Thời gian
    'Severity', 'Start_Time', 'End_Time', #Mức độ nghiêm trọng, Thời gian bắt đầu, Thời gian kết thúc,
    
    # Vị trí
    'Start_Lat', 'Start_Lng', 'Distance(mi)', 'City','State' , # 'Vĩ độ_bắt đầu', 'Kinh độ_bắt đầu', 'Khoảng cách (dặm)', 'Thành phố',
    
    # Thời tiết
    'Temperature(F)', 
    'Humidity(%)', # độ ẩm 
    'Visibility(mi)', #tầm nhìn
    'Precipitation(in)', # lượng mưa 

    'Weather_Condition',#tình trạng thời tiết
    
    # Hạ tầng & Môi trường
    'Junction', 'Traffic_Signal', 'Sunrise_Sunset'
]

# --- 3. LOAD DỮ LIỆU ---
try:
    print(f" Đangload file từ: {file_path}")
    print(" Wait...")
    
    
  
    df = pd.read_csv(file_path, usecols=cols_to_keep)
    

    df['Precipitation(in)'] = df['Precipitation(in)'].fillna(0)

    print(f"Load thành công! Kích thước bảng: {df.shape}")
  
    
  
    display(df.sample(5))

except FileNotFoundError:
    print(" LỖI: Không tìm thấy file 'US_Accidents_March23.csv'!")
```

```python
# --- BẮT ĐẦU XỬ LÝ DỮ LIỆU & FEATURE ENGINEERING (BẢN CHUẨN) ---
print("Đang xử lý dữ liệu thời gian...")

# 1. Chuyển đổi sang định dạng datetime (để máy tính cộng trừ được)
# errors='coerce': Nếu dòng nào lỗi ngày tháng thì biến thành NaT (Not a Time) chứ không báo lỗi dừng chương trình
df['Start_Time'] = pd.to_datetime(df['Start_Time'], errors='coerce')
df['End_Time'] = pd.to_datetime(df['End_Time'], errors='coerce')

# 2. TÁCH CỘT THỜI GIAN 
# Tách Giờ (0-23) -> Để biết sáng/chiều/tối
df['Hour'] = df['Start_Time'].dt.hour

# Tách Ngày (1-31)
df['Day'] = df['Start_Time'].dt.day

# Tách Tháng (1-12) -> QUAN TRỌNG: Để biết mùa (Đông/Hè)
df['Month'] = df['Start_Time'].dt.month

# Tách Năm (2016-2023)
df['Year'] = df['Start_Time'].dt.year

# --- [MỚI THÊM VÀO] Tách Thứ (Weekday) ---
# Giá trị trả về: 0 là Thứ 2, ..., 6 là Chủ Nhật
# Cột này bắt buộc phải có để vẽ biểu đồ Heatmap
df['Weekday'] = df['Start_Time'].dt.weekday

# 3. TẠO CỘT DURATION 
# Công thức: (Kết thúc - Bắt đầu) -> Đổi ra phút
df['Duration'] = (df['End_Time'] - df['Start_Time']).dt.total_seconds() / 60

# --- BƯỚC LỌC DỮ LIỆU NHIỄU ---

# 1. Lọc Duration: 
# Xóa các vụ có thời gian âm hoặc bằng 0 (vô lý)
df = df[df['Duration'] > 0]

# Kiểm tra số lượng vụ quá dài (> 24h)
print(f"Số vụ > 24h trước khi xử lý: {len(df[df['Duration'] > 1440])}")

# Capping: Quy hết các vụ > 1440 phút về 1440 phút
df.loc[df['Duration'] > 1440, 'Duration'] = 1440
print("Đã thực hiện Capping thành công!")
print(f"Giá trị Duration lớn nhất hiện tại: {df['Duration'].max()} phút")

# Xóa cột gốc để tiết kiệm RAM (Vì đã tách hết thông tin ra rồi)
df = df.drop(columns=['Start_Time', 'End_Time'])

# --- KIỂM TRA KẾT QUẢ ---
pd.set_option('display.max_columns', None)

print("--- 5 DÒNG ĐẦU TIÊN CỦA DỮ LIỆU SAU KHI XỬ LÝ ---")
display(df.head(25))

# In thêm kích thước bảng để xem còn bao nhiêu dòng sau khi lọc
print(f"\nKích thước dữ liệu hiện tại: {df.shape}")

print("\nThống kê cột Tháng (Month) xem có đủ từ 1 đến 12 không:")
print(df['Month'].value_counts().sort_index())

print("\nThống kê cột Thứ (Weekday) - 0:Thứ 2 -> 6:CN:")
print(df['Weekday'].value_counts().sort_index())
```

```python
# --- QUAN TRỌNG: TẠO BẢN SAO ---
df_dashboard = df.copy()
print(" Đang xử lý dữ liệu cho Dashboard...")

# 1. Capping Duration (Gán trần 24h = 1440 phút)
# Để biểu đồ không bị dẹt vì mấy vụ tai nạn ảo kéo dài cả năm
df_dashboard.loc[df_dashboard['Duration'] > 1440, 'Duration'] = 1440
df_dashboard = df_dashboard[df_dashboard['Duration'] > 0] # Bỏ lỗi âm

# 2. Điền giá trị thiếu 
# Cột số -> Điền Median
num_cols = ['Temperature(F)', 'Humidity(%)', 'Visibility(mi)', 'Distance(mi)']
for col in num_cols:
    if col in df_dashboard.columns:
        df_dashboard[col] = df_dashboard[col].fillna(df_dashboard[col].median())

# Cột chữ -> Điền Unknown/Mode
if 'City' in df_dashboard.columns:
    df_dashboard['City'] = df_dashboard['City'].fillna('Unknown')

if 'Weather_Condition' in df_dashboard.columns:
    df_dashboard['Weather_Condition'] = df_dashboard['Weather_Condition'].fillna('Không xác định')

print(" Dữ liệu Dashboard đã sạch sẽ!")
```

```python
# 1 & 2. Traffic Signal và Junction: Bỏ qua không cần dịch
# Dữ liệu sẽ tự động giữ nguyên dạng True / False gốc của file

# 3. Sunrise_Sunset (Ngày/Đêm): Giữ nguyên Day/Night
if 'Sunrise_Sunset' in df_dashboard.columns:
    # Chỉ cần điền từ 'Day' vào những ô bị bỏ trống (fillna)
    df_dashboard['Sunrise_Sunset'] = df_dashboard['Sunrise_Sunset'].fillna('Day')

# Hiển thị kết quả để kiểm tra
display(df_dashboard[['City', 'Traffic_Signal', 'Sunrise_Sunset', 'Duration']].head())
```

```python
import os

# --- LƯU TRỮ DỮ LIỆU ĐÃ LÀM SẠCH ---

# 1. Tạo thư mục 'output' (nếu chưa có)
# Bước này quan trọng vì trong ảnh cây thư mục của bạn chưa có folder output
if not os.path.exists('output'):
    os.makedirs('output')
    print(" Đã tạo thư mục 'output'")

# 2. Lưu file Dashboard (Quan trọng nhất để chạy App)
# Lưu ý: encoding='utf-8-sig' để giữ tiếng Việt không bị lỗi font khi mở bằng Excel
# index=False để bỏ cột số thứ tự (0, 1, 2...) thừa thãi
df_dashboard.to_csv('output/dashboard_data.csv', index=False, encoding='utf-8-sig')

print(" Đã lưu xong file tại: output/dashboard_data.csv")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Cấu hình giao diện đẹp
sns.set_style("whitegrid") # Nền trắng có kẻ ô
plt.rcParams['figure.figsize'] = (12, 6) # Kích thước mặc định cho to
plt.rcParams['font.size'] = 12
```

```python
plt.figure(figsize=(8, 5))
# Vẽ biểu đồ cột đếm số lượng
ax = sns.countplot(x='Severity', data=df_dashboard, palette='Reds')

plt.title('Distribution of Accident Severity', fontsize=15)
plt.xlabel('Severity Level (1: Minor -> 4: Severe)')
plt.ylabel('Number of Accidents')

# Hiện số trên đầu cột cho dễ nhìn
for p in ax.patches:
    ax.annotate(f'{int(p.get_height()):,}', (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', xytext=(0, 10), textcoords='offset points')
plt.show()
```

```python
plt.figure(figsize=(12, 6))

# Gom nhóm theo giờ và đếm
accidents_by_hour = df_dashboard.groupby('Hour').size()

# Vẽ biểu đồ đường
sns.lineplot(x=accidents_by_hour.index, y=accidents_by_hour.values, marker='o', color='darkred', linewidth=2.5)

plt.title('Number of Accidents by Hour of Day', fontsize=15)
plt.xlabel('Hour (0 - 23)')
plt.ylabel('Number of Accidents')
plt.xticks(range(0, 24)) # Show all 24 hours
plt.grid(True, linestyle='--')
plt.show()
```

```python
plt.figure(figsize=(14, 7))

# 1. Tạo bảng Pivot (ngang là Giờ, dọc là Thứ)
heatmap_data = df_dashboard.pivot_table(index='Weekday', columns='Hour', values='Severity', aggfunc='count')

# Sắp xếp lại thứ tự từ Thứ 2 đến CN (thay vì 0-6)
days_order = [0, 1, 2, 3, 4, 5, 6] 
heatmap_data = heatmap_data.reindex(days_order)
yticklabels = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

# 2. Vẽ Heatmap
sns.heatmap(heatmap_data, cmap='inferno', annot=False, fmt='d', yticklabels=yticklabels)

plt.title('Heatmap: Accident Frequency by Day of Week and Hour', fontsize=15)
plt.xlabel('Hour of Day')
plt.ylabel('Day of Week')
plt.show()
```

```python
plt.figure(figsize=(12, 6))

# Lấy top 10 thành phố
top_cities = df_dashboard['City'].value_counts().head(10)

# Vẽ biểu đồ cột nằm ngang
sns.barplot(x=top_cities.values, y=top_cities.index, palette='viridis')

plt.title('Top 10 Cities with the Most Accidents', fontsize=15)
plt.xlabel('Number of Accidents')
plt.ylabel('City')
plt.show()
```

```python
plt.figure(figsize=(15, 10))


sample_df = df_dashboard.sample(n=2000000, random_state=42)

# Vẽ Scatter plot (Kinh độ vs Vĩ độ)
sns.scatterplot(x='Start_Lng', y='Start_Lat', data=sample_df, 
                hue='Severity', palette='coolwarm', s=10, alpha=0.5)

plt.title('Map of Accident Distribution', fontsize=15)
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.legend(title='Severity Level', loc='lower right')
plt.show()
```
