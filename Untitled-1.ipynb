```python
import pandas as pd
import os

# --- 1. CẤU HÌNH ĐƯỜNG DẪN ---
file_path = os.path.join('data', 'US_Accidents_March23.csv')

# --- 2. DANH SÁCH CỘT CẦN GIỮ LẠI ---
cols_to_keep = [
    # Mục tiêu & Thời gian
    'Severity', 'Start_Time', 'End_Time', #Mức độ nghiêm trọng, Thời gian bắt đầu, Thời gian kết thúc,
    
    # Vị trí
    'Start_Lat', 'Start_Lng', 'Distance(mi)', 'City', # 'Vĩ độ_bắt đầu', 'Kinh độ_bắt đầu', 'Khoảng cách (dặm)', 'Thành phố',
    
    # Thời tiết
    'Temperature(F)', 
    'Humidity(%)', # độ ẩm 
    'Visibility(mi)', #tầm nhìn
    'Precipitation(in)', # lượng mưa 

    'Weather_Condition',#tình trạng thời tiết
    
    # Hạ tầng & Môi trường
    'Junction', 'Traffic_Signal', 'Sunrise_Sunset'
]

# --- 3. LOAD DỮ LIỆU ---
try:
    print(f" Đangload file từ: {file_path}")
    print(" Wait...")
    
    
  
    df = pd.read_csv(file_path, usecols=cols_to_keep, nrows=2000000)
    

    df['Precipitation(in)'] = df['Precipitation(in)'].fillna(0)

    print(f"Load thành công! Kích thước bảng: {df.shape}")
  
    
  
    display(df.sample(5))

except FileNotFoundError:
    print(" LỖI: Không tìm thấy file 'US_Accidents_March23.csv'!")
```

```python
# --- BẮT ĐẦU XỬ LÝ DỮ LIỆU & FEATURE ENGINEERING (BẢN CHUẨN) ---
print("Đang xử lý dữ liệu thời gian...")

# 1. Chuyển đổi sang định dạng datetime (để máy tính cộng trừ được)
# errors='coerce': Nếu dòng nào lỗi ngày tháng thì biến thành NaT (Not a Time) chứ không báo lỗi dừng chương trình
df['Start_Time'] = pd.to_datetime(df['Start_Time'], errors='coerce')
df['End_Time'] = pd.to_datetime(df['End_Time'], errors='coerce')

# 2. TÁCH CỘT THỜI GIAN 
# Tách Giờ (0-23) -> Để biết sáng/chiều/tối
df['Hour'] = df['Start_Time'].dt.hour

# Tách Ngày (1-31)
df['Day'] = df['Start_Time'].dt.day

# Tách Tháng (1-12) -> QUAN TRỌNG: Để biết mùa (Đông/Hè)
df['Month'] = df['Start_Time'].dt.month

# Tách Năm (2016-2023)
df['Year'] = df['Start_Time'].dt.year


# Cột này bắt buộc phải có để vẽ biểu đồ Heatmap
df['Weekday'] = df['Start_Time'].dt.weekday

# 3. TẠO CỘT DURATION 
# Công thức: (Kết thúc - Bắt đầu) -> Đổi ra phút
df['Duration'] = (df['End_Time'] - df['Start_Time']).dt.total_seconds() / 60

# --- BƯỚC LỌC DỮ LIỆU NHIỄU ---

# 1. Lọc Duration: 
# Xóa các vụ có thời gian âm hoặc bằng 0 (vô lý)
df = df[df['Duration'] > 0]

# Kiểm tra số lượng vụ quá dài (> 24h)
print(f"Số vụ > 24h trước khi xử lý: {len(df[df['Duration'] > 1440])}")

# Capping: Quy hết các vụ > 1440 phút về 1440 phút
df.loc[df['Duration'] > 1440, 'Duration'] = 1440
print("Đã thực hiện Capping thành công!")
print(f"Giá trị Duration lớn nhất hiện tại: {df['Duration'].max()} phút")

# Xóa cột gốc để tiết kiệm RAM (Vì đã tách hết thông tin ra rồi)
df = df.drop(columns=['Start_Time', 'End_Time'])

# --- KIỂM TRA KẾT QUẢ ---
pd.set_option('display.max_columns', None)

print("--- 5 DÒNG ĐẦU TIÊN CỦA DỮ LIỆU SAU KHI XỬ LÝ ---")
display(df.head(50))

# In thêm kích thước bảng để xem còn bao nhiêu dòng sau khi lọc
print(f"\nKích thước dữ liệu hiện tại: {df.shape}")

print("\nThống kê cột Tháng (Month) xem có đủ từ 1 đến 12 không:")
print(df['Month'].value_counts().sort_index())

print("\nThống kê cột Thứ (Weekday) - 0:Thứ 2 -> 6:CN:")
print(df['Weekday'].value_counts().sort_index())
```

```python
# --- QUAN TRỌNG: TẠO BẢN SAO ---
df_dashboard = df.copy()
print(" Đang xử lý dữ liệu cho Dashboard...")

# 1. Capping Duration (Gán trần 24h = 1440 phút)
# Để biểu đồ không bị dẹt vì mấy vụ tai nạn ảo kéo dài cả năm
df_dashboard.loc[df_dashboard['Duration'] > 1440, 'Duration'] = 1440
df_dashboard = df_dashboard[df_dashboard['Duration'] > 0] # Bỏ lỗi âm

# 2. Điền giá trị thiếu 
# Cột số -> Điền Median
num_cols = ['Temperature(F)', 'Humidity(%)', 'Visibility(mi)', 'Distance(mi)']
for col in num_cols:
    if col in df_dashboard.columns:
        df_dashboard[col] = df_dashboard[col].fillna(df_dashboard[col].median())

# Cột chữ -> Điền Unknown/Mode
if 'City' in df_dashboard.columns:
    df_dashboard['City'] = df_dashboard['City'].fillna('Unknown')

if 'Weather_Condition' in df_dashboard.columns:
    df_dashboard['Weather_Condition'] = df_dashboard['Weather_Condition'].fillna('Không xác định')

print(" Dữ liệu Dashboard đã sạch sẽ!")
```

```python


# 1. Dịch Traffic Signal
if 'Traffic_Signal' in df_dashboard.columns:
    df_dashboard['Traffic_Signal'] = df_dashboard['Traffic_Signal'].replace({True: 'Có Đèn', False: 'Không Đèn'})

# 2. Dịch Junction (Giao lộ)
if 'Junction' in df_dashboard.columns:
    df_dashboard['Junction'] = df_dashboard['Junction'].map({True: 'Có giao lộ', False: 'Đường thẳng'})

# 3. Dịch Sunrise_Sunset (Ngày/Đêm)
if 'Sunrise_Sunset' in df_dashboard.columns:
    df_dashboard['Sunrise_Sunset'] = df_dashboard['Sunrise_Sunset'].map({'Day': 'Ban ngày', 'Night': 'Ban đêm'}).fillna('Ban ngày')

display(df_dashboard[['City', 'Traffic_Signal', 'Sunrise_Sunset', 'Duration']].head())
```

```python
# --- BƯỚC 1: HÀM GOM NHÓM THỜI TIẾT (7 NHÓM) ---
def group_weather_7_classes(condition):
    # 1. Xử lý giá trị rỗng
    if pd.isna(condition) or str(condition).strip() == 'nan' or condition == 'Không xác định':
        return 'Unknown/Others'
    
    c = str(condition).lower()
    
    # 2. Ưu tiên bắt các nhóm Nguy hiểm/Đặc thù trước
    
    # Nhóm Tuyết/Băng (Nguy hiểm nhất -> Ưu tiên bắt trước)
    if any(x in c for x in ['snow', 'sleet', 'ice', 'wintry', 'freezing', 'hail']):
        return 'Snow/Ice'
    
    # Nhóm Bão/Sấm sét (Ưu tiên hơn mưa thường)
    elif any(x in c for x in ['thunder', 't-storm', 'tornado']):
        return 'Thunderstorm'
    
    # Nhóm Sương mù/Khói (Ảnh hưởng tầm nhìn)
    elif any(x in c for x in ['fog', 'mist', 'haze', 'smoke', 'dust', 'volcanic']):
        return 'Fog/Mist'
    
    # Nhóm Mưa (Các loại mưa còn lại)
    elif any(x in c for x in ['rain', 'drizzle', 'shower', 'precipitation', 'squall']):
        return 'Rain'
    
    # 3. Các nhóm thời tiết Tốt/Bình thường
    
    # Nhóm Mây (Phổ biến thứ 2)
    elif any(x in c for x in ['cloud', 'overcast']):
        return 'Cloudy'
    
    # Nhóm Trời quang (Phổ biến nhất)
    elif any(x in c for x in ['clear', 'fair']):
        return 'Clear/Fair'
    
    # Còn lại (Gió to mà không mưa, v.v...)
    else:
        return 'Unknown/Others'

# --- BƯỚC 2: ÁP DỤNG VÀO DỮ LIỆU ---
df_dashboard['Weather_Group'] = df_dashboard['Weather_Condition'].apply(group_weather_7_classes)

# Kiểm tra kết quả gom nhóm
print("✅ Đã gom thành 7 nhóm:")
print(df_dashboard['Weather_Group'].value_counts())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

plt.figure(figsize=(12, 7))

# 1. Tính toán bảng chéo (Tỷ lệ %)
weather_severity = pd.crosstab(df_dashboard['Weather_Group'], 
                               df_dashboard['Severity'], 
                               normalize='index') * 100

# Sắp xếp thứ tự nhóm theo ý đồ phân tích (Tốt -> Xấu -> Rất xấu)
custom_order = ['Clear/Fair', 'Cloudy', 'Fog/Mist', 'Rain', 'Thunderstorm', 'Snow/Ice', 'Unknown/Others']
weather_severity = weather_severity.reindex(custom_order)

# 2. Vẽ biểu đồ
ax = weather_severity.plot(kind='bar', stacked=True, figsize=(12, 7), 
                           colormap='RdYlBu_r', edgecolor='black', linewidth=0.5)

plt.title('Tác động của Thời tiết đến Mức độ Nghiêm trọng (7 Nhóm chính)', fontsize=15)
plt.xlabel('Điều kiện thời tiết', fontsize=12)
plt.ylabel('Tỷ lệ phần trăm (%)', fontsize=12)
plt.xticks(rotation=0)
plt.legend(title='Mức độ (Severity)', bbox_to_anchor=(1.02, 1), loc='upper left')

# 3. Hiện số liệu (Chỉ hiện nếu > 2% để đỡ rối)
for p in ax.patches:
    width, height = p.get_width(), p.get_height()
    x, y = p.get_xy() 
    if height > 2: 
        ax.text(x + width/2, 
                y + height/2, 
                f'{height:.1f}%', 
                ha='center', 
                va='center',
                color='white' if height > 10 else 'black',
                fontsize=9,
                fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import geopandas as gpd

# --- KHẮC PHỤC LỖI GEOPANDAS 1.0 ---
# Thay vì dùng gpd.datasets.get_path (đã bị xóa), ta load trực tiếp từ URL GitHub
# Đây là file GeoJSON chứa ranh giới các bang của Mỹ
url = "https://raw.githubusercontent.com/PublicaMundi/MappingAPI/master/data/geojson/us-states.json"

try:
    print("Đang tải bản đồ ranh giới các bang...")
    usa = gpd.read_file(url)
    print("✅ Đã tải xong bản đồ nền!")
except Exception as e:
    print(f"❌ Lỗi tải bản đồ: {e}")
    # Nếu lỗi mạng, dừng chương trình tại đây

# --- BẮT ĐẦU VẼ ---
if 'usa' in locals():
    # 1. Khởi tạo khung hình to đẹp
    fig, ax = plt.subplots(figsize=(20, 12))

    # 2. VẼ LỚP NỀN (BẢN ĐỒ CÁC BANG)
    # color='#f0f0f0': Nền xám nhạt
    # edgecolor='white': Viền trắng (nhìn hiện đại hơn viền đen)
    usa.plot(ax=ax, color='#e0e0e0', edgecolor='white', linewidth=1.5)

    # 3. LẤY MẪU DỮ LIỆU TAI NẠN
    # Lọc bỏ Alaska/Hawaii (để bản đồ tập trung vào đất liền Mỹ)
    # start_lng > -130 và < -65 là kinh độ của phần lục địa Mỹ
    df_map = df_dashboard[
        (df_dashboard['Start_Lng'] > -130) & (df_dashboard['Start_Lng'] < -65) &
        (df_dashboard['Start_Lat'] > 24) & (df_dashboard['Start_Lat'] < 50)
    ]
    
    # Lấy mẫu khoảng 200k điểm (nếu máy khỏe thì tăng lên 500k)
    # Dùng try-except để tránh lỗi nếu dữ liệu ít hơn số cần lấy
    try:
        sample_size = min(200000, len(df_map))
        df_plot = df_map.sample(n=sample_size, random_state=42)
    except:
        df_plot = df_map

    # 4. VẼ CÁC ĐIỂM TAI NẠN CHỒNG LÊN
    sns.scatterplot(
        x='Start_Lng', 
        y='Start_Lat', 
        data=df_plot, 
        hue='Severity', 
        palette='Reds',  # Tông màu đỏ cảnh báo
        s=1,             # Chấm nhỏ xíu để không che bản đồ
        alpha=0.3,       # Độ trong suốt (để thấy chỗ nào đậm là nhiều tai nạn)
        linewidth=0,     # Không viền
        ax=ax            # Vẽ đè lên trục ax cũ
    )

    # 5. TRANG TRÍ
    plt.title(f'Bản đồ Phân bố Tai nạn Giao thông Hoa Kỳ (Mẫu {sample_size:,} điểm)', fontsize=20)
    plt.xlabel('Kinh độ', fontsize=12)
    plt.ylabel('Vĩ độ', fontsize=12)
    
    # Giới hạn khung hình cho vừa khít nước Mỹ lục địa
    plt.xlim(-128, -65)
    plt.ylim(24, 50)
    
    plt.legend(title='Mức độ', markerscale=5, loc='lower right')
    plt.grid(False) # Tắt lưới
    plt.show()
```
