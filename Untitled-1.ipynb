```python
import pandas as pd
import os

# --- 1. CẤU HÌNH ĐƯỜNG DẪN ---
file_path = os.path.join('data', 'US_Accidents_March23.csv')

# --- 2. DANH SÁCH CỘT CẦN GIỮ LẠI ---
cols_to_keep = [
    # Mục tiêu & Thời gian
    'Severity', 'Start_Time', 'End_Time', #Mức độ nghiêm trọng, Thời gian bắt đầu, Thời gian kết thúc,
    
    # Vị trí
    'Start_Lat', 'Start_Lng', 'Distance(mi)', 'City', # 'Vĩ độ_bắt đầu', 'Kinh độ_bắt đầu', 'Khoảng cách (dặm)', 'Thành phố',
    
    # Thời tiết
    'Temperature(F)', 
    'Humidity(%)', # độ ẩm 
    'Visibility(mi)', #tầm nhìn
    'Precipitation(in)', # lượng mưa 

    'Weather_Condition',#tình trạng thời tiết
    
    # Hạ tầng & Môi trường
    'Junction', 'Traffic_Signal', 'Sunrise_Sunset'
]

# --- 3. LOAD DỮ LIỆU ---
try:
    print(f" Đangload file từ: {file_path}")
    print(" Wait...")
    
    
  
    df = pd.read_csv(file_path, usecols=cols_to_keep, nrows=2000000)
    

    df['Precipitation(in)'] = df['Precipitation(in)'].fillna(0)

    print(f"Load thành công! Kích thước bảng: {df.shape}")
  
    
  
    display(df.sample(5))

except FileNotFoundError:
    print(" LỖI: Không tìm thấy file 'US_Accidents_March23.csv'!")
```

```python
# --- BẮT ĐẦU XỬ LÝ DỮ LIỆU & FEATURE ENGINEERING (BẢN CHUẨN) ---
print("Đang xử lý dữ liệu thời gian...")

# 1. Chuyển đổi sang định dạng datetime (để máy tính cộng trừ được)
df['Start_Time'] = pd.to_datetime(df['Start_Time'], errors='coerce')
df['End_Time'] = pd.to_datetime(df['End_Time'], errors='coerce')

# 2. TÁCH CỘT THỜI GIAN 
# Tách Giờ (0-23) -> Để biết sáng/chiều/tối
df['Hour'] = df['Start_Time'].dt.hour

# Tách Ngày (1-31)
df['Day'] = df['Start_Time'].dt.day

# Tách Tháng (1-12) -> QUAN TRỌNG: Để biết mùa (Đông/Hè)
df['Month'] = df['Start_Time'].dt.month

# Tách Năm (2016-2023)
df['Year'] = df['Start_Time'].dt.year

# 3. TẠO CỘT DURATION 
# Công thức: (Kết thúc - Bắt đầu) -> Đổi ra phút
df['Duration'] = (df['End_Time'] - df['Start_Time']).dt.total_seconds() / 60

# --- BƯỚC LỌC DỮ LIỆU NHIỄU ---

# 1. Lọc Duration: 
df = df[df['Duration'] > 0]
print(f"Số vụ > 24h trước khi xử lý: {len(df[df['Duration'] > 1440])}")
df.loc[df['Duration'] > 1440, 'Duration'] = 1440
print("Đã thực hiện Capping thành công!")
print(f"Giá trị Duration lớn nhất hiện tại: {df['Duration'].max()} phút")

df = df.drop(columns=['Start_Time', 'End_Time'])

# --- KIỂM TRA KẾT QUẢ ---
pd.set_option('display.max_columns', None)

print("--- 5 DÒNG ĐẦU TIÊN CỦA DỮ LIỆU SAU KHI XỬ LÝ ---")
display(df.head(25))

# In thêm kích thước bảng để xem còn bao nhiêu dòng sau khi lọc
print(f"\nKích thước dữ liệu hiện tại: {df.shape}")

print("\nThống kê cột Tháng (Month) xem có đủ từ 1 đến 12 không:")
print(df['Month'].value_counts().sort_index())
```

```python
# --- QUAN TRỌNG: TẠO BẢN SAO ---
df_dashboard = df.copy()
print(" Đang xử lý dữ liệu cho Dashboard...")

# 1. Capping Duration (Gán trần 24h = 1440 phút)
# Để biểu đồ không bị dẹt vì mấy vụ tai nạn ảo kéo dài cả năm
df_dashboard.loc[df_dashboard['Duration'] > 1440, 'Duration'] = 1440
df_dashboard = df_dashboard[df_dashboard['Duration'] > 0] # Bỏ lỗi âm

# 2. Điền giá trị thiếu 
# Cột số -> Điền Median
num_cols = ['Temperature(F)', 'Humidity(%)', 'Visibility(mi)', 'Distance(mi)']
for col in num_cols:
    if col in df_dashboard.columns:
        df_dashboard[col] = df_dashboard[col].fillna(df_dashboard[col].median())

# Cột chữ -> Điền Unknown/Mode
if 'City' in df_dashboard.columns:
    df_dashboard['City'] = df_dashboard['City'].fillna('Unknown')

if 'Weather_Condition' in df_dashboard.columns:
    df_dashboard['Weather_Condition'] = df_dashboard['Weather_Condition'].fillna('Không xác định')

print(" Dữ liệu Dashboard đã sạch sẽ!")
```

```python


# 1. Dịch Traffic Signal
if 'Traffic_Signal' in df_dashboard.columns:
    df_dashboard['Traffic_Signal'] = df_dashboard['Traffic_Signal'].replace({True: 'Có Đèn', False: 'Không Đèn'})

# 2. Dịch Junction (Giao lộ)
if 'Junction' in df_dashboard.columns:
    df_dashboard['Junction'] = df_dashboard['Junction'].map({True: 'Có giao lộ', False: 'Đường thẳng'})

# 3. Dịch Sunrise_Sunset (Ngày/Đêm)
if 'Sunrise_Sunset' in df_dashboard.columns:
    df_dashboard['Sunrise_Sunset'] = df_dashboard['Sunrise_Sunset'].map({'Day': 'Ban ngày', 'Night': 'Ban đêm'}).fillna('Ban ngày')

display(df_dashboard[['City', 'Traffic_Signal', 'Sunrise_Sunset', 'Duration']].head())
```

```python
import os

# --- LƯU TRỮ DỮ LIỆU ĐÃ LÀM SẠCH ---

# 1. Tạo thư mục 'output' (nếu chưa có)
# Bước này quan trọng vì trong ảnh cây thư mục của bạn chưa có folder output
if not os.path.exists('output'):
    os.makedirs('output')
    print(" Đã tạo thư mục 'output'")

# 2. Lưu file Dashboard (Quan trọng nhất để chạy App)
# Lưu ý: encoding='utf-8-sig' để giữ tiếng Việt không bị lỗi font khi mở bằng Excel
# index=False để bỏ cột số thứ tự (0, 1, 2...) thừa thãi
df_dashboard.to_csv('output/dashboard_data.csv', index=False, encoding='utf-8-sig')

print(" Đã lưu xong file tại: output/dashboard_data.csv")
```
